# 现在.. 揭晓第一部分

首先，你可以选择性地将一些设置添加到 .env 文件中：

`RUN_EVERY_N_MINUTES=60`
这决定了交易员 Agent 运行的频率，如果未在 .env 文件中指定，默认为每 60 分钟运行一次。

`RUN_EVEN_WHEN_MARKET_IS_CLOSED=False`
这决定了交易员是否在非交易时间仍然运行，如果未在 .env 文件中指定，默认为 False。

`USE_MANY_MODELS=False`
这决定了是否在 OpenAI 之外还使用 DeepSeek、Gemini 和 Grok，通过 DeepSeek API 和 OpenRouter。
如果未在 .env 文件中指定，默认为 False，此时全程使用 gpt-4o-mini。

如果你愿意，请进行这些更改！
