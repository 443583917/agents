# 欢迎来到学习指南！

## 简介

我将这门课程设计为适合各种背景的学习者。对于那些相对新手，我准备了一些技术简报来帮助你建立专业知识。

这些指南设计为自学材料；你可以按照自己的节奏进行学习、研究和实验。

我在编写这些指南时充分借助了 AI 的帮助，并尽量使它们对课程尽可能有用和相关。

课程只有一个要求：足够的耐心！请记住，最好的学习方式之一就是解决问题——如果你对某个难题感到沮丧，记住这正是学习发生的时刻！如果需要帮助，请随时联系我。
