# 解释"内核"

在 notebook 背后运行着一个 Python 进程，当你运行每个单元格时它就会执行。这个 Python 进程被称为内核（Kernel）。每个 notebook 都有自己独立的内核。

你可以点击上面的"Restart Kernel"按钮。

如果你然后尝试运行下一个单元格，你会得到一个错误，因为 favorite_fruit 不再被定义。你需要从 notebook 顶部重新运行所有单元格。然后下一个单元格应该可以正常运行。

````python
print(f"我最喜欢的水果是 {favorite_fruit}")
````

你可以点击工具栏中的"Clear All Outputs"来清理。在将代码推送到 GitHub 等仓库之前清理输出是一个好习惯，否则文件会很大且难以阅读。

````python
spams = ["spam"] * 1000
print(spams)

# 运行后最好清除输出！
````

# 使用 Markdown

那么这些有文字的区域是怎么回事，比如这个？实际上，还有一种不同类型的单元格叫做"Markdown"单元格，用于添加像这样的说明。点击 [+ Markdown] 按钮添加一个新的 Markdown 单元格。

使用 Markdown 格式添加一些注释，可以复制粘贴以下内容：

```
# 这是一级标题
## 这是二级标题
### 这是三级标题

我喜欢 Jupyter Lab 因为它
- 简单
- 灵活
- 令人满意
```

只需在单元格中按 Shift+Return 即可将其变成格式化文本。
如果要删除它，点击单元格并按垃圾桶图标。

# 课程中遇到的小知识

这不一定是 notebook 的特性，但有一个在 notebook 中很有用的好包值得了解。

`tqdm` 包可以在包装任何可迭代对象时打印一个漂亮的进度条。

````python
# 这是一段没有进度条的代码
# 它将运行 10 秒钟，而你不知道发生了什么……

import time

spams = ["spam"] * 1000

for spam in spams:
    time.sleep(0.01)
````

````python
# 现在，加上一个漂亮的小进度条：

import time
from tqdm import tqdm

spams = ["spam"] * 1000

for spam in tqdm(spams):
    time.sleep(0.01)
````

````python
# 换个话题，这是在 markdown 中打印输出的一种有用方式

from IPython.display import Markdown, display

display(Markdown("# 这是一个大标题！\n\n- 这是一个要点\n- 这也是\n- 我也是！"))
````
