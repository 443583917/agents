## 如何开始构建你的想法

所以你已经有了一个想构建的东西。

恭喜！最重要的部分已经完成了。想法就是一切！

但现在你不确定从哪里开始。我被问到很多关于这个的问题。

我想给你三条关键建议。

## 建议 1：做一名科学家

成为 AI 工程师需要戴两顶帽子：软件工程师和数据科学家。

在项目的开始阶段，关键是摘下你的软件工程帽子，坚定地戴上数据科学帽子。

很多来自软件工程背景的人（包括我自己！）都在这方面挣扎。人们常常落入自己的舒适区，问关于架构、设计、数据管道、可扩展性、部署的问题。

这些都是很好的问题，但在早期阶段它们不是关键问题。它们不是困难的问题。它们不会决定你的想法成败。

对于关键问题，你需要成为一名科学家。

问自己：
1. 你将如何评估模型是否成功？你将测量什么？
2. 你有什么数据，你需要什么数据？
3. 你如何构建小型原型来测试不同的技术并了解模型性能？

首先解决科学问题——你能用 LLM 实现什么以及如何实现。

我建议先在 Notebook 中工作，然后再用 Python 模块，这样可以迫使自己以这种科学思维方式运作。

这在 Agent 项目中更加关键。画出一个关于 Agent 如何协作的大型架构图可能很诱人，就像技术图表一样。但这是像软件工程师一样思考。我建议用不同的方式：尝试不同的方法，研究什么有效，测试假设并迭代。

## 建议 2：拥抱研发

好吧，这和建议 1 很相似，但值得重复！

我经常被问到这样的问题："Ed，我想构建以下解决方案。我应该使用模型 A、B 还是 C？我应该使用 Agent、RAG 还是微调？"

我的回答几乎总是：
- 你应该全部尝试！关键是做实验；（a）提出评估标准，（b）开发精心策划的数据集，（c）测试不同的想法看它们表现如何
- 我通常有直觉哪个效果最好，但我的直觉经常是错的！不要相信我：自己试试
- 对于实验，永远没有替代品。

## 建议 3：心怀远大，但从小处着手！

我也经常有学生给我发 200-300 行代码说"Ed，这表现不好，请修复它"。😂

正如我在调试指南中说的——这行不通！

当开始项目时，从小处和简单处着手至关重要。完善每个提示词；在每个步骤上详细工作。让自己确信每次 LLM 调用都在按预期执行，并在输入上迭代直到响应一致且可靠。

如果你总是增量地工作，使用小的、可测试的构建块，你应该对发生的一切完全清晰。

当然，我喜欢帮助人们做项目，帮助解决问题也很棒。但如果你增量地推进项目，你应该始终清楚确切是什么表现不佳——这样我就能非常高效地帮助你。

# 总结

你可能注意到了这 3 条建议中的共同主题。对非数据科学背景的人来说，以这种特别的方式处理问题可能会感到不适；它感觉"随意"和令人不满意。但在我看来，这是成为一名成功的 AI 工程师需要掌握的最重要技能：适应不确定性，享受实验，拥抱做一名科学家。
