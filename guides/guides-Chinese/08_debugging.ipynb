### 调试公式：用 10 行代码重现问题

优秀的调试既有艺术也有科学。掌握它的最好方法就是练习！但以下是基本要点：

1. 重现
持续地重现你的问题

2. 简化
将你的问题简化到最简单的可重现方式。我无法告诉你有多少人给我发了 100 多行代码然后说"Ed，这运行不了，请修复它"。这行不通！我要做的第一件事就是将其简化为核心问题——理想情况下是 10-20 行代码或更少。有一次，我向一个学生展示了如何用 2 行代码重现它！

_关键是，**你**做这件事比任何人都容易。_ 所以虽然这可能很乏味，但这几乎总是最好的开始方式。简化到几行代码。顺便说一句，当你这样做时，你自己往往就已经找到问题了！

3. 寻求帮助
一旦你用几行代码重现了问题，如果你看不出问题所在——你应该寻求帮助！当然值得问 ChatGPT 和它的朋友们；给他们简短的示例和完整的堆栈跟踪。

也请随时联系我！我在这里是为了让你的学习体验尽可能的棒，如果你卡住了，我会帮你脱困。
