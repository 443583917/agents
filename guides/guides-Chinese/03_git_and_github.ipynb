### 贡献你的更改到仓库，与他人分享你的贡献

我非常欢迎你的贡献。这为所有其他学生增加了价值，我自己也很喜欢看到它们！作为额外的好处，你将在 Github 上获得作为仓库贡献者的认可。

以下是发起 PR 的整体步骤和关键说明：
https://edwarddonner.com/pr

提交前请检查：
1. 你的 PR 只包含 community_contributions 中的更改（除非我们讨论过）
2. 所有 notebook 输出清晰
3. 总共不超过 2,000 行代码，文件不要太多
4. 不要包含不必要的测试文件、过于冗长的 README 或 .env.example、emoji 或其他 LLM 痕迹！

非常感谢！
