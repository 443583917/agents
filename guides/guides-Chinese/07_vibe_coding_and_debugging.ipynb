# Vibe Coding 与调试

"Vibe Coding" 是借助 LLM 辅助编程的亲切说法。在 Cursor 中这样做尤其简单和美妙！但有一些好的技巧能让它效果更好，我在课程中会讲到。以下是几个关键要点：

1. 如果你让 ChatGPT 或 Claude 等 LLM 写代码，在提示词中包含当前日期，并指示 LLM 生成使用最新版本 API 的代码。这对 OpenAI API 尤其重要！ChatGPT 经常尝试使用自己 API 的过时版本……

2. 在你的提示词中，要求 LLM 保持代码尽可能简短。LLM 似乎很喜欢添加大量额外的、不必要的检查，这会使代码杂乱且难以发现问题。

3. 向多个 LLM 提出同样的问题，选择最清晰、最简单的答案。

4. 类似地：获取一个 LLM 的答案，让另一个 LLM 验证其正确性以及是否可以更简单或更清晰。

5. 让 LLM 给你提供 3 种解决问题的变体方案，选择最清晰的。

## 针对较大项目的 Vibe Coding

尽量避免让 LLM 一次性生成 100-200 行或更多的代码；这将非常难以调试和找出问题所在（除非你已经很有经验）。

相反：首先将你的问题分解为小的、可独立测试的步骤，每个步骤相对较小。如果你不确定如何分解问题——这是你可以让 LLM 帮你做的事情！

然后对于每个构建块：
- 使用上述技巧让 LLM 构建代码
- 也让 LLM 编写测试来测试和验证代码
- 自己测试并确保其正常工作

这将让你能够自信地构建更大的项目。
