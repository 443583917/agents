## 使用 OpenAI 以外模型的关键背景知识——请先阅读！

整个课程中，我们使用 API 连接地球上最强大的 LLM。

这些 LLM 背后的公司，如 OpenAI、Anthropic、Google 和 DeepSeek，已经构建了 Web 端点。你通过向 Web 地址发起 HTTP 请求并传入所有关于提示词的信息来调用他们的模型。

但如果每次调用 API 都需要构建 HTTP 请求，那会很痛苦。

为了简化这个过程，OpenAI 团队编写了一个名为"Python 客户端库"的 Python 工具，它封装了 HTTP 调用。所以你写 Python 代码，它来调用 Web。

这就是 `openai` 库的本质。

### `openai` Python 客户端库是什么

它是：
- 一个轻量级的 Python 工具
- 将你的 Python 请求转换为 HTTP 调用
- 将 HTTP 调用返回的结果转换为 Python 对象

### 它不是什么

- 它没有任何实际运行大语言模型的代码！没有 GPT 代码！它只是发起 Web 请求
- 没有科学计算代码，也没有专门针对 OpenAI 的特殊内容

### 如何使用：

```python
# 创建一个 OpenAI Python 客户端，用于向 OpenAI 发起 Web 调用
openai = OpenAI()

# 发起调用
response = openai.chat.completions.create(model="gpt-4.1-mini", messages=[{"role":"user", "content": "2+2等于多少？"}])

# 打印结果
print(response.choices[0].message.content)
```

### 这做了什么

当你发起 Python 调用：`openai.chat.completions.create()`
它只是向这个 URL 发起 Web 请求：`https://api.openai.com/v1/chat/completions`
然后将响应转换为 Python 对象。

就这样。

这里是[直接 Web HTTP 调用](https://platform.openai.com/docs/guides/text?api-mode=chat&lang=curl)的 API 文档
这里是使用 [Python 客户端库](https://platform.openai.com/docs/guides/text?api-mode=chat&lang=python)的同一份 API 文档

## 有了这些背景知识——我如何使用其他 LLM？

事实证明——超级简单！

所有其他主流 LLM 都有与 OpenAI 兼容的 API 端点。

OpenAI 帮了大家一个忙：他们说，大家都可以使用我们的工具将 Python 转换为 Web 请求。我们允许你将这个工具从调用 `https://api.openai.com/v1` 改为调用你指定的任何 Web 地址。

因此你可以使用 OpenAI 的工具来调用非 OpenAI 的模型，像这样：

`not_actually_openai = OpenAI(base_url="https://somewhere.completely.different/", api_key="another_providers_key")`

重要的是要理解，这个 OpenAI 代码只是一个用于向端点发起 HTTP 调用的工具。所以即使我们使用 OpenAI 团队的代码，我们也可以用它来调用 OpenAI 以外的模型。

以下是所有主流提供商的 OpenAI 兼容端点。甚至包括在本地使用 Ollama。Ollama 在你本地机器上提供一个端点，而且他们也使其兼容 OpenAI——非常方便。

## 以下是 Gemini、DeepSeek、Ollama 和 OpenRouter 的示例

### 示例 1：使用 Gemini 代替 OpenAI

1. 访问 Google Studio 设置账户：https://aistudio.google.com/
2. 将你的密钥作为 GOOGLE_API_KEY 添加到 `.env` 文件中
3. 同时将其再次添加为 GEMINI_API_KEY 到 `.env` 中——这在后面会有用。

然后：

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv(override=True)

GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
google_api_key = os.getenv("GOOGLE_API_KEY")
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)
response = gemini.chat.completions.create(model="gemini-2.5-flash-lite", messages=[{"role":"user", "content": "2+2等于多少？"}])
print(response.choices[0].message.content)
```

## API 的成本

每次 API 调用的成本确实非常低——我们在课程中使用的大多数模型每次调用只需几分钱。

但极其重要的是需要注意：

1. 一个复杂的 Agentic 项目可能涉及许多 LLM 调用——也许 20-30 次——因此成本会累积起来。设置限制并监控使用情况非常重要。

2. 使用 Agentic AI 时，Agent 有可能陷入循环或执行比预期更多的处理。你应该监控你的 API 使用情况，永远不要充入超过你心理预期的金额。一些 API 有"自动充值"设置，可以自动从你的卡扣款——我强烈建议你关闭这个功能。

3. 你应该只花费你心理上能接受的金额。如果你想，可以使用 Ollama 作为免费的替代方案。DeepSeek、Gemini 2.5 Flash 和 gpt-4.1-nano 也显著更便宜。

请记住，这些 LLM 调用通常涉及数万亿次浮点计算——总要有人付电费的！

### Ollama：付费 API 的免费替代方案

Ollama 是一个在本地机器上运行的产品。它可以运行开源模型，并在你的电脑上提供一个与 OpenAI 兼容的 API 端点。

首先，访问 https://ollama.com 下载 Ollama。

然后在 Cursor 的终端中（View 菜单 >> Terminal），运行此命令下载模型：

```shell
ollama pull llama3.2
```

警告：注意不要使用 llama3.3 或 llama4——这些是大得多的模型，不适合家用电脑。

现在，任何时候我们有这样的代码：
`openai = OpenAI()`
你可以用这个直接替代：
`openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')`
同时将模型名称如 **gpt-4o-mini** 替换为 **llama3.2**。

你不需要在 .env 文件中放任何东西；使用 Ollama，一切都在你的电脑上运行。你没有调用云端的第三方，没有人持有你的信用卡信息，所以不需要密钥！上面代码中的 `api_key='ollama'` 只是因为 OpenAI 客户端库要求传入 api_key，但这个值会被 Ollama 忽略。

完整示例：

```python
# 你需要在电脑上执行一次
!ollama pull llama3.2

from openai import OpenAI
MODEL = "llama3.2"
openai = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")

response = openai.chat.completions.create(
 model=MODEL,
 messages=[{"role": "user", "content": "2 + 2 等于多少？"}]
)

print(response.choices[0].message.content)
```

## OpenAI Agents SDK——特定说明

使用 OpenAI Agents SDK（第 2 周和第 6 周），使用 OpenAI 自身提供的任何模型都特别简单。只需传入模型名称：

`agent = Agent(name="Jokester", instructions="你是一个讲笑话的人", model="gpt-4o-mini")`

你也可以替换成任何具有 OpenAI 兼容 API 的其他提供商。通过以下 3 个步骤实现：

```python
DEEPSEEK_BASE_URL = "https://api.deepseek.com/v1"
deepseek_client = AsyncOpenAI(base_url=DEEPSEEK_BASE_URL, api_key=deepseek_api_key)
deepseek_model = OpenAIChatCompletionsModel(model="deepseek-chat", openai_client=deepseek_client)
```

然后创建 Agent 时直接提供这个模型：

`agent = Agent(name="Jokester", instructions="你是一个讲笑话的人", model=deepseek_model)`

对于任何其他 OpenAI 兼容 API，你也可以使用类似的方法，同样 3 个步骤：

```python
# 额外导入
from agents import OpenAIChatCompletionsModel
from openai import AsyncOpenAI

# 步骤 1：指定提供商提供 OpenAI 兼容 API 的基础 URL 端点
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
GROK_BASE_URL = "https://api.x.ai/v1"
GROQ_BASE_URL = "https://api.groq.com/openai/v1"
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"
OLLAMA_BASE_URL = "http://localhost:11434/v1"

# 步骤 2：为该端点创建 AsyncOpenAI 对象
gemini_client = AsyncOpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)
grok_client = AsyncOpenAI(base_url=GROK_BASE_URL, api_key=grok_api_key)
groq_client = AsyncOpenAI(base_url=GROQ_BASE_URL, api_key=groq_api_key)
openrouter_client = AsyncOpenAI(base_url=OPENROUTER_BASE_URL, api_key=openrouter_api_key)
ollama_client = AsyncOpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")

# 步骤 3：创建模型对象，在创建 Agent 时提供
gemini_model = OpenAIChatCompletionsModel(model="gemini-2.5-flash", openai_client=gemini_client)
grok_3_model = OpenAIChatCompletionsModel(model="grok-3-mini-beta", openai_client=openrouter_client)
llama3_3_model = OpenAIChatCompletionsModel(model="llama-3.3-70b-versatile", openai_client=groq_client)
grok_3_via_openrouter_model = OpenAIChatCompletionsModel(model="x-ai/grok-3-mini-beta", openai_client=openrouter_client)
llama_3_2_local_model = OpenAIChatCompletionsModel(model="llama3.2", openai_client=ollama_client)
```
