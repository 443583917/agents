### 现在来看一些示例：

````python
# 定义一个异步函数

import asyncio

async def do_some_work():
    print("开始工作")
    await asyncio.sleep(1)
    print("工作完成")
````

````python
# 这会做什么？

do_some_work()
````

````python
# OK 让我们再试一次！

await do_some_work()
````

````python
# 这段代码有什么问题？

async def do_a_lot_of_work():
    do_some_work()
    do_some_work()
    do_some_work()

await do_a_lot_of_work()
````

````python
# 有趣的警告！让我们修复它

async def do_a_lot_of_work():
    await do_some_work()
    await do_some_work()
    await do_some_work()

await do_a_lot_of_work()
````

````python
# 现在让我们并行执行
# 重要的是要认识到这不是你可能习惯的"多线程"
# asyncio 库在单线程上运行，但它使用循环在一个任务等待时切换到另一个任务

async def do_a_lot_of_work_in_parallel():
    await asyncio.gather(do_some_work(), do_some_work(), do_some_work())

await do_a_lot_of_work_in_parallel()
````

### 最后——尝试写一个调用 do_a_lot_of_work_in_parallel 的 Python 模块

参见顶部的链接；你的模块中需要类似这样的代码：

```python
if __name__ == "__main__":
    asyncio.run(do_a_lot_of_work_in_parallel())
```
