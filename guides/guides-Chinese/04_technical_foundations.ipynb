## 主题 1：ChatGPT 与 OpenAI API 的区别

### ChatGPT 和 GPT API 有什么区别？两者都由 OpenAI 提供。

#### ChatGPT 是一个终端用户工具。它是一个面向消费者的聊天产品，使用者是 AI 用户。
- 它有免费计划，也有功能更多的付费订阅计划。
- 订阅计划让用户几乎无限制地使用聊天产品。

#### API 是提供给 AI 工程师——软件工程师和数据科学家——用于构建其他商业产品的服务。
- 它允许技术人员（像你我一样）访问底层模型（如"GPT4.1"和"o3"），以便我们可以构建自己的产品。
- 如果我们愿意，我们可以使用 API 构建自己版本的 ChatGPT，并向最终用户收费。
- 与大多数 API 一样，OpenAI 根据 API 使用量收取少量费用。对于课程中使用 gpt-4o-mini 的大多数示例，每次 API 调用的费用大约为 $0.001。

### 我已经每月支付 $20 使用 ChatGPT——为什么还需要为 API 付费？

- 希望现在这一点已经清楚了。API 不是为消费者准备的；它是为工程师构建自己的可收费平台而准备的。
- 如果你能基于订阅获得 API 访问权限，那你就可以以更便宜的价格向他人提供 ChatGPT 工具，让 OpenAI 破产！
- 请记住：每次 API 调用可能需要 10,000,000,000,000 次浮点计算——这些计算消耗电力！

除了调用 API，你也可以在本地运行开源模型，但通常它们的计算量要少 1,000 倍——即使很小，这些处理仍然会计入你的电费账单。
